```python
# Heatmap for correlations
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Between Variables')
plt.show()
```

**Sales are highly correlated with TV advertising costs.**


---

As sales are correlated with only one variable 'TV', we can train our model using linear regression.

```python
X_train, X_test, Y_train, Y_test = train_test_split(df[['TV']], df[['Sales']], test_size = 0.3, random_state=0)
```

```python
print(X_train)
```

```python
print(Y_train)
```

```python
print(X_test)
```

```python
print(Y_test)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
res = model.predict(X_test)
print(res)
```

Proof for model accuracy:

coefficient*X_test value+intercept=model prediction

```python
# Evaluate the model
mse = mean_squared_error(Y_test, res)
r2 = r2_score(Y_test, res)
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

```python
# Proof of model accuracy
warnings.filterwarnings("ignore")
print(f"\nModel Coefficient: {model.coef_[0][0]:.5f}")
print(f"Model Intercept: {model.intercept_[0]:.5f}")
print(f"Example Prediction: {model.coef_[0][0]} * 69.2 + {model.intercept_[0]} = {model.predict([[69.2]])[0][0]}")
```

```python
0.05473199 * 69.2 + 7.14382225
```

```python
plt.figure(figsize=(8, 6))  # Set the figure size
plt.plot(res, color='blue', linestyle='-', linewidth=2, label='Predicted Sales')  # Enhanced plot

# Adding labels and title
plt.title('Predicted Sales vs. Index', fontsize=16, fontweight='bold')
plt.xlabel('Index', fontsize=14)
plt.ylabel('Predicted Sales', fontsize=14)

# Adding gridlines
plt.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.7)

# Adding legend
plt.legend(loc='upper left', fontsize=12)

# Adding an annotation (optional, to highlight a specific point)
plt.annotate(f'Max Prediction: {res.max():.2f}',
             xy=(res.argmax(), res.max()),
             xytext=(res.argmax() + 5, res.max() + 1),
             arrowprops=dict(facecolor='red', arrowstyle='->'),
             fontsize=12, color='darkred')

# Showing the plot
plt.show()
```

```python
# Regression line for training data
plt.scatter(X_train, Y_train, color='blue', label='Training Data')
plt.plot(X_train, model.predict(X_train), color='red', label='Regression Line')
plt.title('Regression Line for Training Data')
plt.xlabel('TV Advertising Costs')
plt.ylabel('Sales')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Regression line for testing data
plt.scatter(X_test, Y_test, color='green', label='Test Data')
plt.plot(X_test, model.predict(X_test), color='red', label='Regression Line')
plt.title('Regression Line for Test Data')
plt.xlabel('TV Advertising Costs')
plt.ylabel('Sales')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Plot the regression line with corrected calculation
line_x = np.linspace(0, 300, 100)  # Generate a range of X values
line_y = model.intercept_[0] + model.coef_[0][0] * line_x  # Ensure scalar values are used

plt.scatter(X_train, Y_train, color='blue', label='Train Data')
plt.scatter(X_test, Y_test, color='green', label='Test Data')
plt.plot(line_x, line_y, color='red', label='Regression Line')  # Plot the regression line

plt.xlim(0, 300)
plt.ylim(0, 25)
plt.xlabel('TV Advertising Cost')
plt.ylabel('Sales')
plt.legend()
plt.title('Regression Line with Train and Test Data')
plt.show()

```

```python
# Residual plot to evaluate model
residuals = Y_test.values.flatten() - res.flatten()
plt.scatter(X_test, residuals, color='purple', alpha=0.7)
plt.axhline(y=0, color='red', linestyle='--', label='Zero Residual Line')
plt.title('Residual Plot')
plt.xlabel('TV Advertising Costs')
plt.ylabel('Residuals (Actual - Predicted)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

**Observations for residual plot:**


---

1. Residuals are fairly randomly distributed around zero, suggesting a good fit.
2. No clear patterns indicate that linear regression is appropriate.

```python
# Predict sales for a new TV ad expense
new_tv_cost = 75
predicted_sales = model.predict([[new_tv_cost]])[0][0]
print(f"\nPredicted Sales for TV Advertising Cost of ${new_tv_cost}: {predicted_sales:.2f}")
```
